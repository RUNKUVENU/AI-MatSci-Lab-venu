```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_features = pd.read_csv('cbfv_features.csv')
```

```python
# 2. Correlation Heatmap
# CBFV generates hundreds of features; select a subset for visual clarity
subset_cols = df_features.columns[2:17] 
corr_matrix = df_features[subset_cols].corr()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', center=0)
plt.title('CBFV Feature Correlation Heatmap (Subset)')
plt.show()
```
