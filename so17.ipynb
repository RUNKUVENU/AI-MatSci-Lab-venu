```python
import pandas as pd
from CBFV import composition
```

```python
df = pd.read_csv('cleaned_cp_dataset.csv')
```

```python
df.head(10)
```

```python
df = df.rename(columns={'Formula': 'formula', 'Cp': 'target'})
```

```python
# 2. Generate the feature vectors
X, y, formulae, skipped = composition.generate_features(
    df, 
    elem_prop='magpie', 
    drop_duplicates=False, 
    extend_features=True
)
```

```python
# 3. Inspect the Matrix
print(f"Feature matrix shape: {X.shape}")
```

```python
# 4. Check for all-zero features and missing values
zero_features = (X == 0).all().sum()
nan_count = X.isna().sum().sum()
```

```python
# 5. Export the validated feature matrix
X.to_csv('cbfv_features.csv', index=False)
print("Feature matrix successfully saved and ready for commit.")
```
